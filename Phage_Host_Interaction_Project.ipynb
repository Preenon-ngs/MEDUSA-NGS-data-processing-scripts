```python
# ============================================================
# Download EMBL Genome from EBI FTP
# ============================================================

import os
import subprocess

# Example accession (E. coli K-12 MG1655 complete genome)
accession = "U00096"

# Updated URL using ENA's API for EMBL download
ftp_url = (
    f"https://www.ebi.ac.uk/ena/browser/api/embl/{accession}?download=true"
)

output_file = f"{accession}.dat.gz"

print("Downloading from EMBL-EBI FTP...")
print(ftp_url)

# Use subprocess.run for better error handling and capture output
result = subprocess.run(f"wget -O {output_file} {ftp_url}", shell=True, capture_output=True, text=True)

if result.returncode != 0:
    print("wget stdout:", result.stdout)
    print("wget stderr:", result.stderr)
    raise Exception(f"Download failed with exit code {result.returncode}. See above for wget output.")


if os.path.exists(output_file):
    if os.path.getsize(output_file) == 0:
        raise Exception(f"Download completed, but '{output_file}' is empty. The remote file might be empty or corrupted.")
    print("Download completed.")
else:
    raise Exception("Download failed. File was not created.")
```

```python
# ============================================================
# Uncompress EMBL File
# ============================================================

import gzip
import shutil
import os

compressed_file_path = f"{accession}.dat.gz"
uncompressed_file_path = f"{accession}.dat"

# Check if downloaded file exists and its size
if not os.path.exists(compressed_file_path):
    raise FileNotFoundError(f"Downloaded file '{compressed_file_path}' not found. Download might have failed.")

file_size = os.path.getsize(compressed_file_path)
print(f"Downloaded file '{compressed_file_path}' size: {file_size} bytes")

if file_size == 0:
    raise ValueError(f"Downloaded file '{compressed_file_path}' is empty. The download might have failed or the remote file was empty.")

# The ENA API with download=true returns the uncompressed file directly,
# even if the output filename ends with .gz. So, we just rename it.
shutil.move(compressed_file_path, uncompressed_file_path)

# Verify the uncompressed file's existence and size
if not os.path.exists(uncompressed_file_path):
    raise FileNotFoundError(f"File '{uncompressed_file_path}' was not created or renamed.")

uncompressed_size = os.path.getsize(uncompressed_file_path)
print(f"Final EMBL file '{uncompressed_file_path}' size: {uncompressed_size} bytes")

if uncompressed_size == 0:
    raise ValueError(f"Final EMBL file '{uncompressed_file_path}' is empty after renaming. The downloaded content might have been empty.")

print("Genome extracted.")
```

```python
# ============================================================
# Extract CDS Using Gene Name
# ============================================================

import os

# Ensure the .dat file exists
file_path = f"{accession}.dat"
if not os.path.exists(file_path):
    raise FileNotFoundError(f"File '{file_path}' not found. Please check the download and uncompression steps.")

# We assume the file is not empty based on the successful execution of the previous cell.
# If parsing fails, it will likely indicate a malformed file rather than an empty one.

# Parse the EMBL file to get the record object
with open(file_path, "r") as handle:
    records = list(SeqIO.parse(handle, "embl"))

if not records:
    raise ValueError(f"No EMBL records found in '{file_path}'. The file might be malformed or not contain valid EMBL entries.")

record = records[0] # Assuming a single genome, take the first record

gene_name = "gyrB"

cds_sequence = None
protein_id = None
product = None

for feature in record.features:

    if feature.type == "CDS":

        genes = feature.qualifiers.get("gene", [])

        if gene_name in genes:

            cds_sequence = feature.extract(record.seq)

            protein_id = feature.qualifiers.get("protein_id", ["NA"])[0]
            product = feature.qualifiers.get("product", ["Unknown"])[0]

            break

if cds_sequence is None:
    raise Exception(f"Gene '{gene_name}' not found in genome.")

print("Gene Found Successfully")
print("Gene        :", gene_name)
print("Protein ID  :", protein_id)
print("Product     :", product)
print("CDS Length  :", len(cds_sequence), "bp")
```

```python
# ============================================================
# Save CDS FASTA
# ============================================================

from Bio.SeqRecord import SeqRecord

cds_record = SeqRecord(
    cds_sequence,
    id=f"{gene_name}_{accession}",
    description=product
)

SeqIO.write(cds_record, "Bacterial_CDS.fasta", "fasta")

print("CDS FASTA saved as Bacterial_CDS.fasta")
```

```python
# ============================================================
# DNA → Protein Translation
# ============================================================

protein_sequence = cds_sequence.translate(to_stop=True)

print("Protein Length:", len(protein_sequence), "AA")

print("\nFirst 100 Amino Acids\n")
print(protein_sequence[:100])

protein_record = SeqRecord(
    protein_sequence,
    id=protein_id,
    description=product
)

SeqIO.write(protein_record, "Protein_Sequence.fasta", "fasta")

print("Protein FASTA saved.")
```

```python
# ============================================================
# Export Summary Table
# ============================================================

import pandas as pd

summary = pd.DataFrame({
    "Parameter":[
        "Genome Accession",
        "Gene Name",
        "Protein ID",
        "Protein Product",
        "CDS Length (bp)",
        "Protein Length (AA)"
    ],
    "Value":[
        accession,
        gene_name,
        protein_id,
        product,
        len(cds_sequence),
        len(protein_sequence)
    ]
})

display(summary)

summary.to_csv("ENA_CDS_Summary.csv", index=False)

print("Summary exported.")
```

```python

# ================================================================
# Import Libraries
# ================================================================

import torch
import numpy as np
import pandas as pd
from Bio import SeqIO

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics.pairwise import cosine_similarity

import umap.umap_ as umap

from transformers import AutoTokenizer, AutoModel

print("PyTorch Version :", torch.__version__)
print("GPU Available   :", torch.cuda.is_available())

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using Device    :", device)
```

```python

# ================================================================
# Read Protein FASTA File
# ================================================================

protein_fasta = "Protein_Sequence.fasta"

records = list(SeqIO.parse(protein_fasta, "fasta"))

protein_ids = []
protein_sequences = []

for record in records:
    protein_ids.append(record.id)
    protein_sequences.append(str(record.seq))

print("Number of Proteins :", len(protein_sequences))

for i in range(min(3, len(protein_sequences))):
    print("\nProtein:", protein_ids[i])
    print("Length :", len(protein_sequences[i]))
    print("Sequence Preview:")
    print(protein_sequences[i][:80], "...")
```

```python

# ================================================================
# Protein Metadata
# ================================================================

metadata = pd.DataFrame({
    "Protein_ID": protein_ids,
    "Sequence_Length": [len(seq) for seq in protein_sequences],
    "Sequence": protein_sequences
})

display(metadata.head())

metadata.to_csv("Protein_Metadata.csv", index=False)

print("Protein metadata exported.")
```

```python

# ================================================================
# Load Meta AI ESM-2 Model
# ================================================================

MODEL_NAME = "facebook/esm2_t33_650M_UR50D"

print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Loading ESM-2 model...")
model = AutoModel.from_pretrained(MODEL_NAME)

model = model.to(device)
model.eval()

print("Model loaded successfully.")
```

```python

# ================================================================
# Generate ESM-2 Embeddings
# ================================================================

esm_embeddings = []

for protein_id, sequence in zip(protein_ids, protein_sequences):

    print("Processing:", protein_id)

    inputs = tokenizer(
        sequence,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    )

    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    embedding = outputs.last_hidden_state.mean(dim=1).squeeze().cpu().numpy()

    esm_embeddings.append(embedding)

esm_embeddings = np.array(esm_embeddings)

print("\nEmbedding Matrix Shape:", esm_embeddings.shape)
```

```python
# ================================================================
# Reload Bacterial Genome FASTA (Required After Runtime Restart)
# ================================================================

from Bio import SeqIO
import pandas as pd
import os

genome_file = "U00096_Genome.fasta"

# Check if the genome FASTA exists
if not os.path.exists(genome_file):
    raise FileNotFoundError(
        f"{genome_file} not found. Make sure the ENA download cell has been executed."
    )

record = SeqIO.read(genome_file, "fasta")

# Create genome_sequence variable (needed in next cell)
genome_sequence = str(record.seq).upper().replace("N", "")

print("Genome loaded successfully.")
print("Genome ID      :", record.id)
print("Genome Length  :", len(genome_sequence), "bp")
print("First 80 bases :", genome_sequence[:80])
```

```python

# ================================================================
# Split Genome into Windows
# ================================================================

WINDOW_SIZE = 512
STRIDE = 256       # Overlapping windows

windows = []
metadata = []

for start in range(0, len(genome_sequence)-WINDOW_SIZE+1, STRIDE):

    fragment = genome_sequence[start:start+WINDOW_SIZE]

    windows.append(fragment)

    metadata.append({
        "Window_ID": len(windows),
        "Start": start,
        "End": start+WINDOW_SIZE
    })

metadata_df = pd.DataFrame(metadata)

print("Total Genome Windows :", len(windows))
display(metadata_df.head())

metadata_df.to_csv(
    "Genome_Window_Metadata.csv",
    index=False
)
```

```python

# ================================================================
# Save Genome Windows FASTA
# ================================================================

from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

records = []

for meta, seq in zip(metadata, windows):

    records.append(
        SeqRecord(
            Seq(seq),
            id=f"Window_{meta['Window_ID']}",
            description=f"{meta['Start']}-{meta['End']}"
        )
    )

SeqIO.write(
    records,
    "Bacterial_Genome_Windows.fasta",
    "fasta"
)

print("Genome windows FASTA created successfully.")
```

```python
# ================================================================
# DNABERT FAST EMBEDDING (10% Genome, Tesla T4)
# Runtime: ~1–2 minutes
# ================================================================
import numpy as np
from tqdm.auto import tqdm
import torch
# -------------------------------
# Use only first 10% of windows
# -------------------------------
sample_fraction = 0.10
num_windows = int(len(windows) * sample_fraction)
# Use only first 500 windows
sample_windows = windows[:500]
BATCH_SIZE = 128
all_embeddings = []
for i in tqdm(range(0, len(sample_windows), BATCH_SIZE)):
    batch_sequences = sample_windows[i:i+BATCH_SIZE]
    encoded = tokenizer(
        batch_sequences,
        padding=True,
        truncation=True,
        max_length=512,
        return_tensors="pt"
    )
    encoded = {k: v.to(device) for k, v in encoded.items()}
    with torch.no_grad():
        outputs = model(**encoded)
        embeddings = outputs.last_hidden_state.mean(dim=1)
    all_embeddings.append(
        embeddings.cpu().numpy()
    )
    # Clear GPU cache after every batch
    torch.cuda.empty_cache()
genome_embeddings = np.concatenate(all_embeddings)
print("Embedding Matrix Shape:", genome_embeddings.shape)
```

```python
# ================================================================
# Save Window Embeddings
# ================================================================

embedding_df = pd.DataFrame(genome_embeddings)

embedding_df.insert(
    0,
    "Window_ID",
    metadata_df["Window_ID"]
)

embedding_df.to_csv(
    "DNABERT_Window_Embeddings.csv",
    index=False
)

np.save(
    "DNABERT_Window_Embeddings.npy",
    genome_embeddings
)

print("Window embeddings saved successfully.")
```

```python

# ================================================================
# Whole Genome Embedding
# ================================================================

whole_embedding = genome_embeddings.mean(axis=0)

print("Genome Embedding Dimension :", whole_embedding.shape)

whole_df = pd.DataFrame(
    whole_embedding.reshape(1,-1)
)

whole_df.insert(
    0,
    "Genome_ID",
    record.id
)

display(whole_df.iloc[:,:12])

whole_df.to_csv(
    "DNABERT_Bacterial_Embeddings.csv",
    index=False
)

torch.save(
    torch.tensor(whole_embedding),
    "DNABERT_Bacterial_Embeddings.pt"
)

np.save(
    "DNABERT_Bacterial_Embeddings.npy",
    whole_embedding
)

print("Whole genome embedding exported.")
```

```python
# ================================================================
# Download Small Bacteriophage Genome (T7)
# ================================================================

import requests
from Bio import SeqIO

# NCBI FASTA URL (Bacteriophage T7, ~39.9 kb)
url = "https://www.ncbi.nlm.nih.gov/sviewer/viewer.fcgi?id=V01146.1&db=nuccore&report=fasta"

response = requests.get(url)

with open("Phage_Genome.fasta", "w") as f:
    f.write(response.text)

record = SeqIO.read("Phage_Genome.fasta", "fasta")

phage_sequence = str(record.seq).upper()

print("Downloaded Phage Genome :", record.id)
print("Genome Length           :", len(phage_sequence), "bp")
```

```python
# ================================================================
# Create Limited Genome Windows
# ================================================================

WINDOW_SIZE = 512
STRIDE = 256

phage_windows = []

for start in range(0, len(phage_sequence)-WINDOW_SIZE+1, STRIDE):

    phage_windows.append(
        phage_sequence[start:start+WINDOW_SIZE]
    )

# Use only first 500 windows (actually T7 has much fewer)
phage_windows = phage_windows[:500]

print("Total Windows Used :", len(phage_windows))
```

```python
# ================================================================
# DNABERT Embeddings for Phage Genome
# ================================================================

import numpy as np
import torch
from tqdm.auto import tqdm

BATCH_SIZE = 64      # Tesla T4 optimized

phage_embeddings = []

for i in tqdm(range(0, len(phage_windows), BATCH_SIZE)):

    batch = phage_windows[i:i+BATCH_SIZE]

    encoded = tokenizer(
        batch,
        padding=True,
        truncation=True,
        max_length=512,
        return_tensors="pt"
    )

    encoded = {k:v.to(device) for k,v in encoded.items()}

    with torch.no_grad():

        outputs = model(**encoded)

        embedding = outputs.last_hidden_state.mean(dim=1)

    phage_embeddings.append(
        embedding.cpu().numpy()
    )

    torch.cuda.empty_cache()

phage_embeddings = np.concatenate(phage_embeddings)

print("Phage Embedding Shape :", phage_embeddings.shape)
```

```python
# ================================================================
# Whole Phage Genome Embedding
# ================================================================

whole_phage_embedding = phage_embeddings.mean(axis=0)

print("Whole Phage Embedding Shape :", whole_phage_embedding.shape)

import pandas as pd

phage_df = pd.DataFrame(
    whole_phage_embedding.reshape(1,-1)
)

phage_df.insert(0,"Genome_ID",record.id)

phage_df.to_csv(
    "DNABERT_Phage_Embedding.csv",
    index=False
)

print("Phage genome embedding exported.")
```

```python
# ================================================================
# Bacterial + Phage Genome Feature Fusion
# ================================================================

import numpy as np
import pandas as pd

# whole_embedding comes from bacterial DNABERT notebook
# whole_phage_embedding comes from Cell 4

genome_fusion = np.concatenate([
    whole_embedding,
    whole_phage_embedding
])

print("Genome Feature Dimension :", genome_fusion.shape)

fusion_df = pd.DataFrame(
    genome_fusion.reshape(1,-1)
)

fusion_df.to_csv(
    "Genome_Feature_Fusion.csv",
    index=False
)

print("Genome fusion saved.")
```

```python
# ================================================================
# Multimodal Genome + Protein Feature Matrix
# ================================================================

import numpy as np
import pandas as pd

# From previous ESM-2 notebook
# whole_embedding          -> Bacterial DNABERT (768)
# whole_phage_embedding    -> Phage DNABERT (768)
# esm_embeddings           -> Bacterial protein embeddings (1280)

# Average protein embeddings
protein_embedding = esm_embeddings.mean(axis=0)

print("Protein Embedding Shape :", protein_embedding.shape)

# Multimodal feature vector
multimodal_vector = np.concatenate([
    whole_embedding,
    whole_phage_embedding,
    protein_embedding
])

print("Multimodal Feature Size :", multimodal_vector.shape)

multimodal_df = pd.DataFrame(
    multimodal_vector.reshape(1,-1)
)

multimodal_df.insert(0,"Sample","Ecoli_T7")

multimodal_df.to_csv(
    "Multimodal_Feature_Matrix.csv",
    index=False
)

print("Multimodal feature matrix created.")
```

```python
# ================================================================
# PyTorch Geometric Node Features
# ================================================================

import numpy as np
import pandas as pd
import torch
import os

# ------------------------------------------------
# Load bacterial DNABERT embedding
# ------------------------------------------------
if os.path.exists("DNABERT_Bacterial_Embedding_10Percent.npy"):
    whole_embedding = np.load("DNABERT_Bacterial_Embedding_10Percent.npy")
elif os.path.exists("DNABERT_Bacterial_Embeddings.npy"):
    whole_embedding = np.load("DNABERT_Bacterial_Embeddings.npy")
else:
    raise FileNotFoundError("Bacterial DNABERT embedding (.npy) not found.")

# ------------------------------------------------
# Load phage DNABERT embedding
# ------------------------------------------------
if os.path.exists("DNABERT_Phage_Embedding.npy"):
    whole_phage_embedding = np.load("DNABERT_Phage_Embedding.npy")
elif os.path.exists("DNABERT_Phage_Embeddings.npy"):
    whole_phage_embedding = np.load("DNABERT_Phage_Embeddings.npy")
else:
    # If only CSV exists
    phage_df = pd.read_csv("DNABERT_Phage_Embedding.csv")
    whole_phage_embedding = phage_df.iloc[0,1:].values.astype(np.float32)

# ------------------------------------------------
# Create protein embedding from ESM-2 output
# ------------------------------------------------
# esm_embeddings was generated in the previous notebook
protein_embedding = esm_embeddings.mean(axis=0)

# ------------------------------------------------
# Check dimensions
# ------------------------------------------------
print("Bacterial DNABERT :", whole_embedding.shape)
print("Phage DNABERT     :", whole_phage_embedding.shape)
print("ESM2 Protein      :", protein_embedding.shape)

# Keep genome embeddings at 768 dimensions
whole_embedding = whole_embedding[:768]
whole_phage_embedding = whole_phage_embedding[:768]

# Convert ESM2 (1280) → 768
protein_embedding_768 = protein_embedding[:768]

# ------------------------------------------------
# Stack node features
# ------------------------------------------------
node_features = torch.tensor(
    np.vstack([
        whole_embedding,
        whole_phage_embedding,
        protein_embedding_768
    ]),
    dtype=torch.float32
)

print("\nNode Feature Tensor Shape:", node_features.shape)

# Save for PyTorch Geometric
torch.save(node_features, "PyG_Node_Features.pt")

print("Saved: PyG_Node_Features.pt")
```
